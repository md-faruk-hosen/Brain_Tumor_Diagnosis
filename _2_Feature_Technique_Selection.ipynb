```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import numpy as np
```

```python
data = {
    "Feature_Selection": ["SHAP", "PCA", "Lasso", "mRMR"],
    "ACC": [0.9790, 0.9191, 0.8932, 0.8909],
    "AUC": [0.9857, 0.9891, 0.9841, 0.9837],
    "PRE": [0.9066, 0.9164, 0.8895, 0.8893],
    "SP": [0.9029, 0.9125, 0.8851, 0.8820],
    "SN": [0.9040, 0.9138, 0.8866, 0.8843],
    "F1": [0.8795, 0.8917, 0.8569, 0.8541],
    "MCC": [0.9790, 0.9191, 0.8932, 0.8909]
}
```

```python
df = pd.DataFrame(data)
```

```python
# Reset matplotlib settings to default
plt.rcParams.update(plt.rcParamsDefault)

# Set Times New Roman font with fallback
plt.rcParams['font.family'] = ['Times New Roman', 'serif']
plt.rcParams['mathtext.fontset'] = 'stix'

# Set global plot style with large font sizes
plt.rcParams['font.size'] = 32
plt.rcParams['axes.labelsize'] = 36
plt.rcParams['axes.titlesize'] = 40
plt.rcParams['xtick.labelsize'] = 30
plt.rcParams['ytick.labelsize'] = 30
plt.rcParams['legend.fontsize'] = 28
plt.rcParams['figure.dpi'] = 1000
plt.rcParams['savefig.dpi'] = 1000
plt.rcParams['figure.facecolor'] = 'white'
```

```python
# Enhanced color palette for Feature Selection techniques
# feature_selection_palette = {
#     'SHAP': '#4C72B0',    # Blue
#     'PCA': '#55A868',     # Green
#     'Lasso': '#C44E52',   # Red
#     'mRMR': '#8172B3'     # Purple
# }

feature_selection_palette = {
    'SHAP': '#1F77B4',    # Bright Blue
    'PCA': '#2CA02C',     # Vivid Green
    'Lasso': '#D62728',   # Bold Red
    'mRMR': '#FF7F0E'     # Rich Orange
}

# feature_selection_palette = {
#     'SHAP': '#FF6F61',    # Coral Red
#     'PCA': '#6A0572',     # Deep Purple
#     'Lasso': '#0EAD69',   # Emerald Green
#     'mRMR': '#FFAA33'     # Golden Amber
# }
```

```python
# Metric palette
metric_palettes = {
    'ACC': 'viridis',
    'AUC': 'plasma',
    'PRE': 'magma',
    'SP': 'cividis',
    'SN': 'rocket',
    'F1': 'mako',
    'MCC': 'flare'
}
```

```python
# Function to save figures in both PNG and PDF in a dedicated folder
def save_figure(fig, filename):
    # Create the output folder if it doesn't exist
    output_folder = "Experiment_Output_Figures/Feature_Selection"
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # Full paths for both formats
    png_path = os.path.join(output_folder, f"{filename}.png")
    pdf_path = os.path.join(output_folder, f"{filename}.pdf")
    
    # Save with tight layout and white background - both formats
    fig.savefig(png_path, dpi=1000, bbox_inches='tight', facecolor='white', format='png')
    fig.savefig(pdf_path, dpi=1000, bbox_inches='tight', facecolor='white', format='pdf')
    
    print(f"Saved: {png_path} and {pdf_path}")
    plt.show()
    plt.close(fig)
```

```python
# Function to generate the violin plot visualization
def generate_violin_plot():
    metrics = ['ACC', 'AUC', 'PRE', 'SP', 'SN', 'F1', 'MCC']
    
    # Prepare data in long format for seaborn
    melted_df = pd.melt(df, id_vars=['Feature_Selection'], 
                        value_vars=metrics, 
                        var_name='Metric', 
                        value_name='Score')
    
    # Create a violin plot showing distribution of scores across metrics for each technique
    fig, ax = plt.subplots(figsize=(20, 14))
    
    # Create the violin plot without black points
    sns.violinplot(x='Feature_Selection', y='Score', data=melted_df,
                   hue='Feature_Selection', palette=feature_selection_palette,
                   inner='box', linewidth=2, ax=ax, legend=False)

    # Customize the plot
    ax.set_title('Distribution of Performance Scores by Feature Selection Technique', 
                fontsize=44, pad=30)
    ax.set_xlabel('Feature Selection Technique', fontsize=40, labelpad=25)
    ax.set_ylabel('Score Distribution Across Metrics', fontsize=40, labelpad=25)
    
    # Set y-axis limits for better focus on the relevant range
    ax.set_ylim(0.7, 1.0)
    
    # Add grid lines for better readability
    # ax.grid(axis='y', linestyle='--', alpha=0.7)
    
    # Save the figure
    save_figure(fig, "feature_selection_violin_plot")
```

```python
metrics = ['ACC', 'AUC', 'PRE', 'SP', 'SN', 'F1', 'MCC']
```

```python
# 1. Bar chart comparing all feature selection techniques across metrics
# Get the data in a format suitable for a grouped bar chart
melted_df = pd.melt(df, id_vars=['Feature_Selection'], 
                    value_vars=metrics, 
                    var_name='Metric', 
                    value_name='Score')

# Sort by Feature_Selection to ensure SHAP (the best) appears first
melted_df = melted_df.sort_values(by=['Feature_Selection'], 
                                  key=lambda x: pd.Categorical(x, ['SHAP', 'PCA', 'Lasso', 'mRMR']))

# Create a grouped bar chart
fig, ax = plt.subplots(figsize=(20, 14))

# Use seaborn to create a grouped bar chart
sns.barplot(x='Metric', y='Score', hue='Feature_Selection', 
            data=melted_df, palette=feature_selection_palette, 
            ax=ax, edgecolor='none')

# Customize the plot
ax.set_title('Comparison of Feature Selection Techniques across Metrics', fontsize=44, pad=30)
ax.set_xlabel('Evaluation Metric', fontsize=40, labelpad=25)
ax.set_ylabel('Score', fontsize=40, labelpad=25)
ax.set_ylim(0.7, 1.0)  # Set y-axis to focus on the relevant range

# Create a more prominent legend
ax.legend(title='Feature Selection', title_fontsize=36, fontsize=34, 
          bbox_to_anchor=(1.05, 1), loc='upper left')

# Add grid lines for better readability
# ax.grid(axis='y', linestyle='--', alpha=0.7)

# Save the figure
save_figure(fig, "feature_selection_comparison_grouped_bar")
```

```python
# 2. Radar/Spider Chart for each Feature Selection Technique
# Prepare the data
categories = metrics
N = len(categories)

# Compute the angle for each metric
angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]  # Close the loop

# Create the figure
fig, ax = plt.subplots(figsize=(16, 16), subplot_kw=dict(polar=True))

# Add lines for each feature selection technique
for i, selection in enumerate(df['Feature_Selection']):
    values = df.loc[i, metrics].values.tolist()
    values += values[:1]  # Close the loop
    
    # Plot the line
    ax.plot(angles, values, linewidth=4, label=selection, 
            color=feature_selection_palette[selection])
    ax.fill(angles, values, alpha=0.1, color=feature_selection_palette[selection])

# Fix axis to go from 0.7 to 1.0 (or adjust as needed)
ax.set_ylim(0.7, 1.0)

# Set category labels
plt.xticks(angles[:-1], categories, fontsize=36)

# Add legend
ax.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1), fontsize=34)

# Add title
plt.title('Feature Selection Techniques Comparison (Radar Chart)', 
          fontsize=44, pad=40, y=1.08)

# Save the figure
save_figure(fig, "feature_selection_radar_chart")
```

```python
# 3. Heatmap showing all scores
# Create a pivot table for the heatmap
pivot_df = df.set_index('Feature_Selection')

# Create the heatmap
fig, ax = plt.subplots(figsize=(18, 12))

# Use seaborn to create the heatmap
heatmap = sns.heatmap(pivot_df, annot=True, fmt=".4f", cmap="YlGnBu", 
                      linewidths=0.5, linecolor='white', annot_kws={'size': 26})

# Set colorbar label size
cbar = heatmap.collections[0].colorbar
cbar.set_label('Score', size=34)
cbar.ax.tick_params(labelsize=28)

# Set title and labels
ax.set_title('Feature Selection Techniques Performance Heatmap', 
             fontsize=40, pad=30)
ax.set_xlabel('Evaluation Metrics', fontsize=36, labelpad=25)
ax.set_ylabel('Feature Selection Technique', fontsize=36, labelpad=25)

# Save the figure
save_figure(fig, "feature_selection_heatmap")
```

```python
# 4. Individual line charts for each metric comparing the techniques
for metric in metrics:
    fig, ax = plt.subplots(figsize=(14, 10))
    
    # Sort data by metric value in descending order for this visualization
    sorted_df = df.sort_values(by=metric, ascending=False)
    
    # Create a bar chart
    bars = ax.bar(sorted_df['Feature_Selection'], sorted_df[metric], 
                 color=[feature_selection_palette[fs] for fs in sorted_df['Feature_Selection']], width=0.6)
    
    
    # Set title and labels
    ax.set_title(f'{metric} Scores by Feature Selection Technique', 
                fontsize=40, pad=30)
    ax.set_xlabel('Feature Selection Technique', fontsize=36, labelpad=25)
    ax.set_ylabel(f'{metric} Score', fontsize=36, labelpad=25)
    
    # Set y-axis limits for better focus on the differences
    ax.set_ylim(min(sorted_df[metric]) - 0.05, 1.02)
    
    # Add grid lines for the y-axis
    # ax.grid(axis='y', linestyle='--', alpha=0.7)
    
    # Save the figure
    save_figure(fig, f"feature_selection_{metric}_comparison")
```

```python
# 5. Create a summary bar chart showing the average performance
# Calculate average performance across all metrics
df['Average'] = df[metrics].mean(axis=1)

# Sort by average performance
sorted_avg_df = df.sort_values(by='Average', ascending=False)

# Create a horizontal bar chart for average performance
fig, ax = plt.subplots(figsize=(16, 10))

# Create horizontal bars
bars = ax.barh(sorted_avg_df['Feature_Selection'], sorted_avg_df['Average'], 
              color=[feature_selection_palette[fs] for fs in sorted_avg_df['Feature_Selection']], 
              height=0.6)

# Set title and labels
ax.set_title('Average Performance of Feature Selection Techniques', 
            fontsize=40, pad=30)
ax.set_xlabel('Average Score Across All Metrics', fontsize=36, labelpad=25)
ax.set_ylabel('Feature Selection Technique', fontsize=36, labelpad=25)

# Set x-axis limits
ax.set_xlim(0.7, 1.0)

# Add grid lines
# ax.grid(axis='x', linestyle='--', alpha=0.7)

# Save the figure
save_figure(fig, "feature_selection_average_performance")
```
