```python
# Reset matplotlib settings to default
mpl.rcParams.update(mpl.rcParamsDefault)

plt.rcParams['font.family'] = ['Times New Roman', 'serif']
plt.rcParams['mathtext.fontset'] = 'stix'  # Mathematical font that matches Times New Roman

# Seaborn should respect matplotlib's rcParams
sns.set_style({'font.family': ['Times New Roman', 'serif']})

# Set global plot style with Times New Roman and much larger font sizes
plt.rcParams['font.size'] = 32  # Substantially increased base font size
plt.rcParams['axes.labelsize'] = 36  # Larger axis labels
plt.rcParams['axes.titlesize'] = 40  # Larger title 
plt.rcParams['xtick.labelsize'] = 30  # Larger tick labels
plt.rcParams['ytick.labelsize'] = 30  # Larger tick labels
plt.rcParams['legend.fontsize'] = 28  # Larger legend
plt.rcParams['figure.dpi'] = 1000  # Lower for testing
plt.rcParams['savefig.dpi'] = 1000  # Lower for testing
plt.rcParams['figure.facecolor'] = 'white'
```

```python
# Enhanced color palettes
metric_palettes = {
    'ACC': 'viridis',
    'AUC': 'plasma',
    'PRE': 'magma',
    'SP': 'cividis',
    'SN': 'rocket',
    'F1': 'mako',
    'MCC': 'flare'
}

classifier_palette = {
    'SVM': '#1F77B4',      # Bright Blue
    'CatBoost': '#FF7F0E', # Rich Orange
    'LightGBM': '#D62728', # Bold Red
    'AttCNN': '#2CA02C',   # Vivid Green
    'AttGRU': '#9467BD'    # Deep Purple
}
```

```python
# Function to save figures in both PNG and PDF in a dedicated folder
def save_figure(fig, filename):
    # Create the output folder if it doesn't exist
    output_folder = "Experiment_Output_Figures/Model_Comparison"
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    # Full paths for both formats
    png_path = os.path.join(output_folder, f"{filename}.png")
    pdf_path = os.path.join(output_folder, f"{filename}.pdf")
    
    # Save with tight layout and white background - both formats
    fig.savefig(png_path, dpi=1000, bbox_inches='tight', facecolor='white', format='png')
    fig.savefig(pdf_path, dpi=1000, bbox_inches='tight', facecolor='white', format='pdf')
    
    print(f"Saved: {png_path} and {pdf_path}")
    plt.show()
    plt.close(fig)
```

```python
# 1. Generate a test bar plot
metric = 'ACC'
fig, ax = plt.subplots(figsize=(18, 12))
sns.barplot(data=df, x="Feature Extractor", y=metric, hue="Classifier", 
            palette=classifier_palette, ax=ax, edgecolor='none')
ax.set_title(f"{metric} Comparison by Feature Extractor and Classifier", pad=30)
ax.set_xlabel("Feature Extractor", labelpad=25)
ax.set_ylabel(metric, labelpad=25)
ax.legend(title="Classifier", title_fontsize=32, fontsize=28, bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

save_figure(fig, f"{metric}_comparison_test")
```

```python
# 2. Generate a test heatmap
pivot_df = df.pivot(index='Feature Extractor', columns='Classifier', values=metric)
fig, ax = plt.subplots(figsize=(16, 12))
heatmap = sns.heatmap(pivot_df, annot=True, fmt=".3f", cmap=metric_palettes[metric], 
            cbar_kws={'label': metric}, ax=ax, 
            linewidths=0.5, linecolor='white',             annot_kws={'size': 26})

# Set colorbar label size separately
cbar = heatmap.collections[0].colorbar
cbar.set_label(metric, size=20)
cbar.ax.tick_params(labelsize=18)

ax.set_title(f"{metric} Scores Heatmap", pad=30, fontsize=26)
ax.set_xlabel("Classifier", fontsize=24, labelpad=25)
ax.set_ylabel("Feature Extractor", fontsize=24, labelpad=25)
plt.tight_layout()

save_figure(fig, f"{metric}_heatmap_test")
```

```python
# 3. Generate a test violin plot
fig, ax = plt.subplots(figsize=(18, 12))
sns.violinplot(data=df, x="Feature Extractor", y=metric, hue="Feature Extractor",
              palette=metric_palettes[metric], ax=ax, legend=False)
ax.set_title(f"{metric} Distribution by Feature Extractor", fontsize=40, pad=30)
ax.set_xlabel("Feature Extractor", fontsize=36, labelpad=25)
ax.set_ylabel(metric, fontsize=36, labelpad=25)
save_figure(fig, f"violin_plot_{metric}_test")
```

```python
metrics = ['ACC', 'AUC', 'PRE', 'SP', 'SN', 'F1', 'MCC']
```

```python
# 1. Generate individual metric comparison plots
for metric in metrics:
    fig, ax = plt.subplots(figsize=(18, 12))
    sns.barplot(data=df, x="Feature Extractor", y=metric, hue="Classifier", 
                palette=classifier_palette, ax=ax, edgecolor='none')
    ax.set_title(f"{metric} Comparison by Feature Extractor and Classifier", pad=30)
    ax.set_xlabel("Feature Extractor", labelpad=25)
    ax.set_ylabel(metric, labelpad=25)
    ax.legend(title="Classifier", title_fontsize=32, fontsize=28, bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    
    save_figure(fig, f"{metric}_comparison")
```

```python
# 2. Heatmaps for all metrics
for metric in metrics:
    pivot_df = df.pivot(index='Feature Extractor', columns='Classifier', values=metric)
    fig, ax = plt.subplots(figsize=(16, 12))
    
    heatmap = sns.heatmap(pivot_df, annot=True, fmt=".3f", cmap=metric_palettes[metric], 
                cbar_kws={'label': metric}, ax=ax, 
                linewidths=0.5, linecolor='white',         annot_kws={'size': 26})
    
    cbar = heatmap.collections[0].colorbar
    cbar.set_label(metric, size=34)
    cbar.ax.tick_params(labelsize=28)
    
    ax.set_title(f"{metric} Scores Heatmap", pad=30, fontsize=40)
    ax.set_xlabel("Classifier", fontsize=36, labelpad=25)
    ax.set_ylabel("Feature Extractor", fontsize=36, labelpad=25)
    plt.tight_layout()
    
    save_figure(fig, f"{metric}_heatmap")
```

```python
# 3. Violin plots
for metric in metrics:
    fig, ax = plt.subplots(figsize=(18, 12))
    sns.violinplot(data=df, x="Feature Extractor", y=metric, hue="Feature Extractor",
                  palette=metric_palettes[metric], ax=ax, legend=False)
    ax.set_title(f"{metric} Distribution by Feature Extractor", fontsize=40, pad=30)
    ax.set_xlabel("Feature Extractor", fontsize=36, labelpad=25)
    ax.set_ylabel(metric, fontsize=36, labelpad=25)
    
    save_figure(fig, f"violin_plot_{metric}")
```

```python
# 4. Combined evaluation metrics plot
    melted_df = pd.melt(df, id_vars=['Feature Extractor', 'Classifier'], 
                        value_vars=metrics, var_name='Metric', value_name='Score')
    
    mean_scores = melted_df.groupby(['Classifier', 'Metric'])['Score'].mean().reset_index()
    
    fig, ax = plt.subplots(figsize=(24, 14))
    sns.barplot(data=mean_scores, x='Metric', y='Score', hue='Classifier',
               palette=classifier_palette, edgecolor='none', ax=ax)
    
    ax.set_title('Combined Evaluation Metrics by Classifier', fontsize=44, pad=35)
    ax.set_xlabel('Evaluation Metric', fontsize=40, labelpad=25)
    ax.set_ylabel('Average Score', fontsize=40, labelpad=25)
    ax.legend(title='Classifier', title_fontsize=36, fontsize=34, 
              bbox_to_anchor=(1.05, 1), loc='upper left')
    
    plt.xticks(ha='right')
    plt.tight_layout()
    
    save_figure(fig, "combined_metrics_by_classifier")
```
